```python
import sympy as sp

# --------------------------------------------------
# Euler-Bernoulli Beam Element (2-node Hermite beam)
# Symbolic derivation of stiffness and mass matrices
# --------------------------------------------------

# Symbols
x, L = sp.symbols('x L', positive=True, real=True)
EI, rho, A = sp.symbols('EI rho A', positive=True, real=True)
xi = x / L

# --------------------------------------------------
# Shape functions (cubic Hermite)
# --------------------------------------------------
N1 = 1 - 3*xi**2 + 2*xi**3
N2 = L*(xi - 2*xi**2 + xi**3)
N3 = 3*xi**2 - 2*xi**3
N4 = L*(-xi**2 + xi**3)

N = sp.Matrix([[N1, N2, N3, N4]])

print('\nShape function vector N(x):')
sp.pprint(N)

# --------------------------------------------------
# Curvature matrix B = d2N/dx2
# --------------------------------------------------
B = N.diff(x, 2)
B = sp.simplify(B)

print('\nCurvature matrix B = d2N/dx2:')
sp.pprint(B)

# --------------------------------------------------
# Stiffness matrix K = ∫ EI * B^T B dx
# --------------------------------------------------
K = sp.integrate(EI * (B.T * B), (x, 0, L))
K = sp.simplify(K)

print('\nStiffness matrix K:')
sp.pprint(K)

K_norm = sp.simplify(K / (EI / L**3))
print('\nNormalized stiffness matrix K / (EI/L^3):')
sp.pprint(K_norm)

# --------------------------------------------------
# Mass matrix M = ∫ rho*A * N^T N dx
# --------------------------------------------------
M = sp.integrate(rho * A * (N.T * N), (x, 0, L))
M = sp.simplify(M)

print('\nMass matrix M:')
sp.pprint(M)

M_norm = sp.simplify(M / (rho * A * L / 420))
print('\nNormalized mass matrix M / (rho*A*L/420):')
sp.pprint(M_norm)

# --------------------------------------------------
# Optional numeric example
# --------------------------------------------------
vals = {
    L: 2.0,
    EI: 1000.0,
    rho: 7800.0,
    A: 0.01
}

K_num = K.subs(vals).evalf()
M_num = M.subs(vals).evalf()

print('\nNumeric example:')
print('K =')
sp.pprint(K_num)
print('\nM =')
sp.pprint(M_num)

# --------------------------------------------------
# Export LaTeX expressions (optional)
# --------------------------------------------------
print('\nLaTeX form of K:')
print(sp.latex(K))

print('\nLaTeX form of M:')
print(sp.latex(M))
```
